```python
def plot_example(df, page, dpi=300, save_path=None):
    # Do not overwrite 'page' column if already present
    if 'page' not in df.columns:
        df = df.copy()
        df['page'] = df.groupby('file_name').ngroup()
    group = df[df['page'] == page]
    file_name = group['file_name'].iloc[0]
    print(f"File name for page {page}: {file_name}")

    img = cv2.imread(file_name)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    for i, row in group.iterrows():
        if i%5==1 or i%5==0:
            x, y, x2, y2 = int(row['x']), int(row['y']), int(row['x2']), int(row['y2'])
            cv2.rectangle(img, (x, y), (x2, y2), color=(255, 0, 0), thickness=2)

    plt.figure(figsize=(10, 10), dpi=dpi)
    plt.imshow(img)
    plt.axis('off')
    if save_path is not None:
        plt.savefig(save_path, bbox_inches='tight', pad_inches=0)
        print(f"Saved to {save_path}")
    plt.show()

file = squares_file

df=pd.read_csv(f"{source_path}\\outputs\\preprocessed_data\\{file}")
running = 'new-laptop'
saved = 'old-laptop'
df=file_IO.change_filename_from_to(df, fr=saved, to=running)
plot_example(df, page=202, dpi=1200,save_path="sample_images_squares.png")
```
