```python
kind='patches_224' 
input_contrastive=file_IO.load_preprocessed_files(kind, mode='contrastive')
input_train=file_IO.load_preprocessed_files(kind, mode='train')
input_val=file_IO.load_preprocessed_files(kind, mode='val')
input_test=file_IO.load_preprocessed_files(kind, mode='test')
file_dict = {'contrastive': input_contrastive, 'train': input_train, 'val': input_val, 'test': input_test}
print('Loaded files:', file_dict)
```

```python
save_root = os.path.join(source_path, 'outputs', 'data_on_disk')

for mode, input_file in file_dict.items():
    print(f"Processing {mode} data...")
    df = pd.read_csv(source_path+f'\\outputs\\preprocessed_data\\'+input_file)
    for idx, row in df.iterrows():
        #print(f"Row {idx}: {row.to_dict()}")
        root = file_IO.assemble_file_name(row['file_name'])
        image = Image.open(row['file_name']).convert("RGB")
        patch = image.crop((row['x'], row['y'], row['x2'], row['y2'] ))
        file_IO.access_or_create_dir(os.path.join(save_root,root))
        base_name = f"{row['x']}_{row['y']}.png"
        patch.save(os.path.join(save_root,root, base_name))
        '''if idx > 5:
            print(f"Processed {idx} rows, breaking for demonstration purposes.")
            break'''
        if idx % 250 == 0:
            print(f"Processed {idx} out of {len(df)} rows for {mode} data.")
    
```
